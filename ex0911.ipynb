```python
import pandas as pd
import matplotlib.pyplot as plt
# 차트에 한글 깨지지 않도록 차트 폰트변경
plt.rcParams['font.family'] = 'Malgun Gothic'
# 차트에 "-"문자 깨짐 방지
plt.rcParams['axes.unicode_minus'] = False
```

```python
# 막대 차트를 그리기 위해 필요한 데이터셋
df = pd.Series([7, 10], index=["M", "F"])
df.plot(kind="bar")

# 실제 제공되는 데이터 셋
gender = pd.Series([
    "M", "F", "M", "M", "F",
    "M", "F", "M", "M", "F",
    "M", "F", "M", "M", "F",
    "M", "M"
])

s = gender.value_counts()
s

print(gender)

# 데이터셋 변경 후 차트를 그려야 한다.
gender.value_counts().plot(kind="bar")
```
