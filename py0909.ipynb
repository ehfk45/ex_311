```python
import pandas as pd 
```

```python
# {레이블 : [], 레이블 : []}
df1 = pd.DataFrame({'이름': ['민수','지영'],'점수': [90, 85]})

# print () 터미널 출력
# display() 노트북 셀에 최적화 출력
df1
```

```python
# index값 행레이블(레이블 인덱스), columns값 열레이블(레이블 인덱스)
# 생략하면 위치값으로 기본값 설정된다.
df2 = pd.DataFrame([['민수', 90],['지영', 85]], index=['첫번째', '두번째'], columns=['이름', '점수'])

df2
```

```python
df3 = pd.DataFrame([
    {'이름' : '민수', '점수' : 90},
    {'이름' : '지영', '점수' : 100},
])
df3
```
