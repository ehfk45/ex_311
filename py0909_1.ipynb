```python
# loc[행옵션, 열옵션] : 레이블 인덱스
# iloc[행옵션, 열옵션] : 위치 인덱스
# [] : 열 레이블, 표현식(T/F 조건식, 논리식,슬라이싱, ....)
```

```python
import pandas as pd

df = pd.DataFrame({
    '이름': ['김철수', '이영희', '박민수', '최수진', '정우진'],
    '학년': [1, 2, 1, 3, 2],
    '점수': [85, 92, 78, 95, 88]
}, index=['A', 'B', 'C', 'D', 'E'])

print(df)
```

```python
df.loc['B', :] # df.loc['B',:]
```

```python
df.loc['B','점수']
```

```python
df.iloc[0,2] # 인덱스 레이블이 숫자로 되어있다면
```

```python
df['이름'] # 열 이름
```

```python
df['A':'A'] # 식-> 지정 식 행 전체
```

```python
df['이름' : '이름']
```

```python
df[['이름', '학년']]
```

```python
df[0 : 5]
```

```python
df['A' : 'E']
```

```python
# [열이름], [열이름리스트], -> 열선택
# [식] -> 행선택
# [:]
# [조건식]
```
