```python
import pandas as pd

df = pd.DataFrame({
    '이름': ['김철수', '이영희', '박민수', '최수진', '정우진'],
    '학년': [1, 2, 1, 3, 2],
    '점수': [85, 92, 78, 95, 88],
    '출석': [90, 95, 80, 98, 85]
})

print(df)
```

```python
df ['이름']
```

```python
# 새로운 열 = 시리즈 타입 
# = 시리즈 타입 리스트 -> 시리즈 변경후 대입 
df['성적'] = ['B', 'A', 'C', 'A+', 'B+'] 
print(df)
```

```python
df['생년'] = 2000 # = 값 -> 같은 값의 시리즈로 변경후 대입(입력)_
```

```python
df
```

```python
df['가입년도'] = [2010, 2009, 2011] # 시리즈로 변경하는데, 갯수가 안맞아서 에러가 발생 

```

```python
df['총점'] = df['점수'] + df['출석']

display(df)
```

```python
# 수정
# [] : 특정 행전체, 열전체 수정 
# iloc[], loc[] : 특정 행, 열 수정
```

```python
df['생년'] = 1999 # 1999 스칼라값이 Series로 변경되어서 대입됨
```

```python
df
```

```python
df[0 : 2] # 0~1행 전체 출력
```

```python
df.iloc[2,4] = 'C+'

df
```

```python
df.loc [2, '성적'] = 'C'

df
```

```python
df.loc[[2, 4], '출석'] = 88
df
```

```python
df.iloc[[2, 4], 3] = 85
df
```

```python
df ['점수'] > 90
```

```python
df.loc [df ['점수'] > 90, '점수'] = 99 # 조건식 매개값 자리는 boolean Series가 들어가야함
# df.loc[pd.Series([false, True, False, True, False]), '점수'] = 99
df
```

```python
# 삭제 drop() : 행, 열 삭제 (메서드)
# 특정 위치 값들을 삭제하는건 불가능 -> 값을 공백, 0, None, NaN 등으로 변경하는 것
# 행들을 선택해서 삭제 - 세로축으로 선택 : axis = 0
# 열들을 선택해서 삭제 - 가로축으로 선택 : axis = 1 
```

```python
df.drop ('생년', axis=1)
```

```python
df.drop (0, axis=0)
```

```python
import pandas as pd

df = pd.DataFrame({
    '이름': ['김철수', '이영희', '박민수', '최수진', '정우진'],
    '학년': [1, 2, 1, 3, 2],
    '점수': [85, 92, 78, 95, 88],
    '출석': [90, 95, 80, 98, 85]
})

print(df)
```

```python
df.describe() #수치형 변수의 모든 통계량, 문자형변수 범주형 인지
```

```python
df['점수'].describe()
```

```python
df['이름'].describe()
```

```python
df['학년'].describe() # 학년 수치형이기때문에 수치기술 통계랑이 출력 -> 수치지만 범주타입 변경 필요
```

```python
df["성적"] = ['B', 'A', 'C', 'A', 'A']  
df
```

```python
df["성적"].describe()
```

```python
# 학년도 수치형이지만 범주형으로 변경 후 통계량 출력
# df['학년'].astype('category') # 데이터를 추출해서 탐색
df['학년'] = df['학년'].astype('category')
df.info()
```

```python
df['학년'].describe()
```

```python
print(df.count())

print(df.sum(numeric_only=True))

print(df.mean(numeric_only=True))

print(df.median(numeric_only=True))

print(df.min(numeric_only=True))

print(df.max(numeric_only=True))

print(df.std(numeric_only=True))

print(df.std(numeric_only=True))

print(df['점수'].max() - df['점수'].min())
```

```python
df.isnull()

# 데이터 행이나 열이 엄청 많으면 사용 불가능
# 데이터가 작은 경우에만 사용
```

```python
df.isnull().sum() # boolean타입의 sum()은 False : 0 , True : 1 로 계산
```

```python
df.duplicated() # 행의 중복이 있으면 그행은 True 
```

```python
df['학년'].value_counts() # 개별 값의 카운팅
```
