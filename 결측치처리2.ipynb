```python
import pandas as pd

data = {
    "날짜": [
        "2025-01-01", "2025-01-02", "2025-01-03",
        "2025-02-01", "2025-02-02", "2025-02-03",
        "2025-03-01", "2025-03-02", "2025-03-03"
    ],
    "매출": [
        100, 120, None,
        200, None, 220,
        300, 320, None
    ]
}

df = pd.DataFrame(data)

df["날짜"] = pd.to_datetime(df["날짜"])

print(df)
```

```python
df['월'] = df['날짜'].dt.month
df
# 시계열 결측치의 경우 이전행, 다음행, (이전행+다음행)/2, 시계열 범주의 평균or중앙값
```

```python
cond = (df['월'] == 1)
print(cond)
```

```python
df[df['월'] == 1]
```

```python
mean1 = (df.loc[df['월'] == 1, '매출']).mean()
mean1
```

```python
mean2 = (df.loc[df['월'] == 2, '매출']).mean()
mean2
```

```python
mean3 = (df.loc[df['월'] == 3, '매출']).mean()
mean3
```

```python
cond = (df['매출'].isna()) & (df['월'] == 1 )
# df[cond]['매출']
df.loc[cond, '매출'] = mean1
df
```

```python
cond = (df['매출'].isna()) & (df['월'] == 2 )
# df[cond]['매출']
df.loc[cond, '매출'] = mean2

```

```python
from sklearn.linear_model import LinearRegression

# 열x, 열y => y = ax + b 
# 열x1, 열x2, 열y -> y = ax1 + ax2 + b 
# 10 = 5 + 3*1 + 2*NaN => 결측치 -> 1
```

```python
import pandas as pd


data = {
    "공부시간": [1, 2, 3, 4, 5, 6, 7, 8],
    "게임시간": [5, 4, 4, 3, 3, 2, 2, 1],
    "시험점수": [45, 50, None, 62, 68, None, 80, 88]
}

df = pd.DataFrame(data)

df
```

```python
train = df[df['시험점수'].notna()]
train
```

```python
X_train = train [['공부시간', '게임시간']]
y_train = train ['시험점수']
```

```python
X_train
```

```python
y_train

```

```python

model = LinearRegression()
model.fit(X_train, y_train)
print(model.coef_)
print(model.intercept_)
```

```python
cond = df['시험점수'].isna()
m = df.loc[cond, ['공부시간', '게임시간']]
m 
```

```python
result_y = model.predict(m)
print(result_y, type(result_y)) # 내부적 사용하는 numpy 가 결과값으로 ... numpy 학습이 필요하다
```

```python
df.loc[cond, '시험점수'] = result_y.round(1)
df 
```
