```python
import pandas as pd

df = pd.DataFrame({
    "이름": ["철수", "영희"],
    "국어": [80, 90],
    "영어": [70, 95],
    "수학": [90, 85]
})

display(df)
```

```python
df_melt = df.melt(
    id_vars="이름",       # 그대로 유지할 열, 이름은 각 행의 대상을 구분하기 위해 그대로 놔둔다
    var_name="과목",      # 기존 열 이름을 저장할 열
    value_name="점수"     # 기존 열의 값을 저장할 열
)

display(df_melt)
```

```python
df_pivot = df_melt.pivot(index = '이름', columns = '과목', values = '점수')
df_pivot
```

```python
display(df)
```

```python
result = df_pivot.reset_index()
result
```
