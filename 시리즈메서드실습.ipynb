```python
# 구간별 값을 매핑하는 메서드 X
# apply(함수) 메서드 
# val 시리즈가 아니다. val.isna() X, pd.isna(값) 함수 사용
def 매출_등급(val) :
    if pd.isna(val):
        return '결측치'
    elif val >= 1000:
        return '상'
    elif val >= 800:
        return '중'
    else:
        return '하'
등급 = 매출.apply(매출_등급)
등급
```

```python
s1 = pd.Series([1, 2])
s2 = pd.Series([1, 2])
s3 = s1.copy() # copy 메서드
s4 = s1 # 대입연산자

display(s1)
display(s2)
display(s3)
display(s4)
```

```python
display(s2.equals(s1))
display(s3.equals(s1))
display(s4.equals(s1))
```

```python
s1[0] = 7
s1
# s1 : 7, 2
# s2 : 0, 2
# s3 : 0, 2
# s4 : 0, 2

display(s2.equals(s1)) # False
display(s3.equals(s1)) # copy() 깊은 복사(deep copy)
display(s4.equals(s1)) # = 얕은 복사(shallow copy)
```
