```python
s = pd.Series([60, 62, 65, 67, 68, 70, 72, 75, 78, 95, 200])
print(s)
print(type(s))
```

```python
# IQR = Q3 - Q1
# 중앙값 = Q2

Q1 = s.quantile(0.25)
Q3 = s.quantile(0.75)
median = s.median()
print('Q1:',Q1,'median:', median,'Q3:', Q3)

IQR = Q3 - Q1 
print ('IQR:',IQR)

# 상한/하한
upper = Q3 + (IQR*1.5)
lower = Q1 - (IQR*1.5)
print ('upper:', upper)
print ('lower:', lower)
```
